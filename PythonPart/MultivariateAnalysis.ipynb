```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("whitegrid")
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
plt.style.use('dark_background')
```

```python
df = pd.read_csv('cleaned_dataset3.csv')
```

```python
pd.crosstab([df['Weather_Conditions'], df['Light_Conditions']],
            df['Accident_Severity'],
            normalize='index').plot(kind='bar', stacked=True, figsize=(12,6))
plt.title("Severity Distribution by Weather and Light Conditions", fontsize=14)
plt.xlabel("Weather and Light Condition Combination", fontsize=12)
plt.ylabel("Proportion of Accidents", fontsize=12)
plt.xticks(rotation=90)
plt.legend(title="Accident Severity")
plt.show()
```

### Insight:

- Across all weather and light condition combinations, the majority of accidents are slight, while serious and fatal accidents make up a much smaller proportion.

### Conclusion:
- There are no major changes in accident severity patterns across different combinations,

```python
plt.figure(figsize=(9,5))
sns.barplot(x='Road_Type', y='Speed_limit', hue='Accident_Severity', data=df, ci=None)
plt.title("Accident Severity by Road Type and Speed Limit", fontsize=14, fontweight='bold')
plt.xlabel("Road Type", fontsize=12)
plt.ylabel("Average Speed Limit (mph)", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### Insight:

- The average speed limit varies by road type, with slip roads and dual carriageways having the highest speed limits, while one-way streets have the lowest.
- Across all road types, the differences between serious, slight, and fatal accidents are small, but higher-speed roads (like dual carriageways and slip roads) tend to show more severe accidents.
- In contrast, one-way streets and roundabouts, which have lower speed limits, are associated with less severe accidents.

### Conclusion:

Roads with higher speed limits such as dual carriageways and slip roads are more prone to severe accidents, while low-speed roads like one-way streets and roundabouts are generally safer.

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='Urban_or_Rural_Area', y='Speed_limit', hue='Accident_Severity', data=df)
plt.title("Accident Severity by Area Type and Speed Limit", fontsize=14, fontweight='bold')
plt.xlabel("Area Type", fontsize=12)
plt.ylabel("Speed Limit (mph)", fontsize=12)
plt.show()
```

### Insight:

- In urban areas, most accidents occur at a consistent speed limit of around 30 mph, with very little variation.

- In rural areas, the speed limits are much higher (40–70 mph) and show greater variation, indicating a wider range of driving speeds.

- Accidents that happen at higher speed limits (mostly in rural areas) tend to have greater severity — including more serious and fatal cases.

- Outliers in rural regions suggest that a few extremely high-speed or low-speed zones may contribute to rare but severe accidents.

### Conclusion:

Urban roads are generally safer due to lower and more controlled speed limits, leading to fewer severe accidents.
In contrast, rural areas, with higher and more variable speed limits, experience more serious and fatal accidents
