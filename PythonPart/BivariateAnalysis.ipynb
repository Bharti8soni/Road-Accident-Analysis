```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style("whitegrid")
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
plt.style.use('dark_background')
```

```python
df = pd.read_csv('final_cleaned.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Road_Surface_Conditions', hue='Accident_Severity', palette='viridis' )
plt.title("Accident Severity by Road Surface Conditions", fontsize=14, fontweight='bold')
plt.xlabel("Road Surface Conditions", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### Insight :
- Most accidents happen on dry roads, and a large number of them are slight accidents.
- Accidents on wet, icy, or snowy roads are very few compared to dry surfaces.
### Conclusion :
- Since most travel happens on dry roads, they record the highest number of accidents.
- Accidents on wet or icy roads are less frequent but still need caution because slippery surfaces can increase risk.

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Road_Type', hue='Accident_Severity')
plt.title('Accident Severity across Road Types', fontsize=14, fontweight='bold')
plt.xlabel('Road Type', fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Light_Conditions', hue='Accident_Severity', data=df)
plt.title("Accident Severity by Light Conditions", fontsize=14, fontweight='bold')
plt.xlabel("Light Conditions", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### Insight :
- Most accidents happen during daylight, followed by darkness with street lights lit.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Accident_Severity', y='Speed_limit', palette='viridis')
plt.title('Speed Limit across Accident Severity', fontsize=14, fontweight='bold')
plt.xlabel('Accident Severity')
plt.ylabel('Speed Limit')
plt.show()
```

#### Slight Accidents:
- Occur mostly at lower speed limits, with a median around 30.
- The interquartile range (IQR) is between approximately 30 and 40.
- There are several outliers at higher speed limits (60 and 70) and one at a very low limit (around 10), indicating slight accidents can still happen at these extremes but are less common.

#### Serious Accidents:
- The median speed limit is higher than for slight accidents, sitting around 30-40.
- The IQR is broader, ranging from 30 to 60, showing that serious accidents occur across a wider range of speed limits.
- The whiskers extend from 20 to 70, covering almost the entire range of speed limits shown.

#### Fatal Accidents:
- Have the highest median speed limit, at 50.
- The IQR is from 30 to 60, similar to serious accidents but with a higher median.
- Like serious accidents, they occur across a wide range of speed limits (20 to 70).

```python
# Extract hour safely from Time column
def extract_hour(time_str):
    try:
        return int(str(time_str).split(':')[0])
    except:
        return np.nan

df['Hour'] = df['Time'].apply(extract_hour)

# Categorize into Time of Day
def get_time_of_day(hour):
    if pd.isna(hour):
        return 'Unknown'
    elif 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_of_Day'] = df['Hour'].apply(get_time_of_day)

# Verify creation
df[['Time', 'Hour', 'Time_of_Day']].head()
```

```python
df['Time_of_Day'].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Time_of_Day', hue='Accident_Severity', data=df, palette ='inferno')
plt.title("Accident Severity by Time of Day", fontsize=14, fontweight='bold')
plt.xlabel("Time of Day", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.show()
```

### Insight:

- Most accidents happen in the afternoon, followed by the morning and evening.
- The fewest accidents occur at night, likely because of lower traffic volume.
- Across all time periods, slight accidents are the most common, while fatal accidents remain very low.

### Conclusion:

Accidents are more frequent during the daytime, especially in the afternoon, when road traffic is highest.

```python
# Function to assign seasons based on month
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    elif month in [9, 10, 11]:
        return 'Autumn'
    else:
        return 'Unknown'

# Apply the function
df['Season'] = df['Month'].apply(get_season)

# Verify it worked
df[['Month', 'Season']].head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Season', hue='Accident_Severity', data=df)
plt.title("Accident Severity by Season", fontsize=14, fontweight='bold')
plt.xlabel("Season", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.show()
```

### Insight
Accidents occur throughout all seasons, with Autumn showing the highest number of accidents, followed closely by Summer and Spring.
Winter has slightly fewer accidents.

### Conclusion:

The number of accidents is fairly consistent across seasons
