### Showing the total number of missing values in each column.

```python
df.isnull().sum()
```

###  Calculating the percentage of missing values in each column and sorts them in descending order to identify columns with the most missing data.

```python
(df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
```

### Checking number of duplicate records

```python
df.duplicated().sum()
```

### Removing duplicates records

```python
df.drop_duplicates(inplace = True)
```

```python
df['Carriageway_Hazards']
```

```python
df['Carriageway_Hazards'].value_counts()
```

###  Removes the column 'Carriageway_Hazards' from the dataset permanently since it contains too many missing values.

```python
df.drop(columns=['Carriageway_Hazards'], inplace=True)
```

```python
df.shape
```

#### Missing values are filled using mode for categorical columns.

```python
for col in ['Weather_Conditions', 'Road_Type', 'Road_Surface_Conditions']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
### Filling with mode
```

```python
df['Time'].fillna(df['Time'].mode()[0], inplace=True)
```

```python
### Rechecks the percentage of missing values in each column after dropping
```

```python
(df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
```
